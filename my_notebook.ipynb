```python
# FEATURE ENGINEERING
# Length of Stay
df['Check_In_Date'] = pd.to_datetime(df['Check_In_Date'])
df['Check_Out_Date'] = pd.to_datetime(df['Check_Out_Date'])

df['Length of stay'] = (
    df['Check_Out_Date'] - df['Check_In_Date']
).dt.days
```

```python
# Booking Lead Time
df['Booking_Created_At'] = pd.to_datetime(
    df['Booking_Created_At']
)

df['Lead time'] = (
    df['Check_In_Date'] - df['Booking_Created_At']
).dt.days
```

```python
# Preparing Target Variable
df = df[df['Booking_Status'].isin(['Abandoned', 'Confirmed'])].copy()

df['Target'] = df['Booking_Status'].map({
    'Abandoned': 1,
    'Confirmed': 0
})
```
